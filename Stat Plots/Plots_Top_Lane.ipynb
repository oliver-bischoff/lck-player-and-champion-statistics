```python
import numpy as np
import matplotlib.pyplot as plt

# Toplane Winrate Daten
winrate = np.array([
    74.3, 75.0, 49.0, 36.1, 54.6, 39.1, 27.0, 62.7, 47.4, 38.7, 61.4,
    63.2, 67.4, 71.4, 60.2, 32.5, 53.0, 30.1, 31.5, 41.9, 25.9,
    79.7, 63.9, 67.7, 43.3, 52.0, 37.4, 53.9, 19.5, 30.9, 21.0,
    71.9, 64.6, 63.0, 55.7, 39.3, 45.3, 38.6, 39.2, 23.9, 53.2,
])

print("Toplane - Winrate")
print()
print("Min:", winrate.min())
print("Max:", winrate.max())
print("Mean:", winrate.mean())
print()

mu = np.mean(winrate)
std = np.std(winrate)

def logistic_raw(x, mu, std, k=1.0):
    z = (x - mu) / std
    return 1 / (1 + np.exp(-k * z))

def winrate_logistic_score(x, mu, std, k=1.0):
    # Rohwerte
    y = logistic_raw(x, mu, std, k)
    
    # Grenzen bei 0% und 100%
    y_min = logistic_raw(0, mu, std, k)
    y_max = logistic_raw(100, mu, std, k)
    
    # Auf 1–10 skalieren
    score = 1 + 9 * (y - y_min) / (y_max - y_min)
    return score

for wr in range(0, 101, 10):
    print(f"{wr}% -> {winrate_logistic_score(wr, mu, std, k=1.0):.1f}") # k-Wert ändern für flacherere vs steilerer Wertverteilungen
```
