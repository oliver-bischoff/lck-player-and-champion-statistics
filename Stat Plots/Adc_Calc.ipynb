```python
##### ---------- Parameter ----------
k = 1.0  # Steilheit, für alle Features anpassbar

# Boundaries
boundaries = {
    "winrate": (0, 100),  # set
    "kda": (0, 8),        # set | open
    "kp": (0, 100),       # set
    "gd15": (-500, 500),  # open
    "csd15": (-10, 10),   # open
    "exp15": (-400, 400), # open
    "dmg": (0, 100),      # set 
    "dpm": (0, 900),      # set | open
    "pog": (0, 200),      # set | open
    "rating": (1, 10),    # set 
}

# ---------- Logistic Raw Function ----------
def logistic_raw(x, mu, std, k):
    z = (x - mu) / std
    return 1 / (1 + np.exp(-k * z))

# ---------- Logistic Score Function Factory ----------
def make_logistic_score_func(data_array, k=1.0, lower_bound=0, upper_bound=100):
    mu = np.mean(data_array)
    std = np.std(data_array)
    
    def score_func(x):
        x_clamped = np.clip(x, lower_bound, upper_bound) # damit alle Werte unter/über den Boundaries auch 1 bzw. 10 ist
        
        y = logistic_raw(x_clamped, mu, std, k)
        y_min = logistic_raw(lower_bound, mu, std, k)
        y_max = logistic_raw(upper_bound, mu, std, k)
        return 1 + 9 * (y - y_min) / (y_max - y_min)
    
    return score_func

# ---------- Wrapped Score Functions ----------
winrate_score_func = make_logistic_score_func(winrate, k=k, lower_bound=boundaries["winrate"][0], upper_bound=boundaries["winrate"][1])
kda_score_func     = make_logistic_score_func(kda, k=k, lower_bound=boundaries["kda"][0], upper_bound=boundaries["kda"][1])
kp_score_func      = make_logistic_score_func(kp, k=k, lower_bound=boundaries["kp"][0], upper_bound=boundaries["kp"][1])
gd15_score_func    = make_logistic_score_func(gd15, k=k, lower_bound=boundaries["gd15"][0], upper_bound=boundaries["gd15"][1])
csd15_score_func   = make_logistic_score_func(csd15, k=k, lower_bound=boundaries["csd15"][0], upper_bound=boundaries["csd15"][1])
exp15_score_func   = make_logistic_score_func(exp15, k=k, lower_bound=boundaries["exp15"][0], upper_bound=boundaries["exp15"][1])
dmg_score_func     = make_logistic_score_func(dmg, k=k, lower_bound=boundaries["dmg"][0], upper_bound=boundaries["dmg"][1])
dpm_score_func     = make_logistic_score_func(dpm, k=k, lower_bound=boundaries["dpm"][0], upper_bound=boundaries["dpm"][1])
pog_score_func     = make_logistic_score_func(pog, k=k, lower_bound=boundaries["pog"][0], upper_bound=boundaries["pog"][1])
rating_score_func  = make_logistic_score_func(rating, k=k, lower_bound=boundaries["rating"][0], upper_bound=boundaries["rating"][1])

# ---------- Gewichtungen ----------
weights = {
    "winrate": 2,
    "kda": 3,
    "kp": 1,
    "gd15": 2,
    "csd15": 1,
    "exp15": 1,
    "dmg": 2,
    "dpm": 3,
    #"pog": 5,
    "rating": 5,
}

# ---------- Overlay Funktion ----------
def calculate_player_score(player_features):
    total_score = 0.0
    
    total_score += weights["winrate"]   * winrate_score_func(player_features["winrate"])
    total_score += weights["kda"]       * kda_score_func(player_features["kda"])
    total_score += weights["kp"]        * kp_score_func(player_features["kp"])
    total_score += weights["gd15"]      * gd15_score_func(player_features["gd15"])
    total_score += weights["csd15"]     * csd15_score_func(player_features["csd15"])
    total_score += weights["exp15"]     * exp15_score_func(player_features["exp15"])
    total_score += weights["dmg"]       * dmg_score_func(player_features["dmg"])
    total_score += weights["dpm"]       * dpm_score_func(player_features["dpm"])
    #total_score += weights["pog"]       * pog_score_func(player_features["pog"])
    total_score += weights["rating"]    * rating_score_func(player_features["rating"])
    
    # Auf 0-100 normieren
    max_total = sum(weights.values()) * 10
    total_score = total_score / max_total * 100
    
    # Vollgerundet
    return round(total_score, 0)

# ---------- Beispiel ----------
input_string = input("Stats eingeben (mit Leerzeichen getrennt): ")
input_string = input_string.replace("\t", " ")  # Tabs aus Excel
input_string = input_string.replace(",", ".")   # Dezimalkomma -> Punkt

values = list(map(float, input_string.split()))

player_stats = {
    "winrate": values[0],
    "kda": values[1],
    "kp": values[2],
    "gd15": values[3],
    "csd15": values[4],
    "exp15": values[5],
    "dmg": values[6],
    "dpm": values[7],
    #"pog": values[8],
    "rating": values[8]
}

overall_score = calculate_player_score(player_stats)
print("Gesamtscore:", overall_score)
```
